Final Insights

```python
#Final Insights
```

Content Performance:
The analysis shows that video and text posts have higher average engagement rates than image and reel posts. Among post categories, Food has the highest average engagement rate, followed by Tech and Lifestyle, while Travel has the lowest average engagement rate. This suggests that the type and category of content are associated with differences in audience engagement.

User Trends:
The dataset contains users from 10 countries and includes different age groups, genders, and device types. Mobile, desktop, and tablet users are all represented in the dataset. The analysis of age and engagement shows very little correlation between age and engagement rate, suggesting that age alone does not strongly explain engagement differences in this dataset.

Behavioral Insights:
Device type shows some differences in user behaviour. Mobile users have the highest average watch time compared with desktop and tablet users. Verified accounts also show a slightly higher average engagement rate than non-verified accounts. These results suggest that device usage and account status may be associated with differences in engagement behaviour.

Sentiment Analysis:
Positive sentiment is the most common sentiment in the dataset, followed by neutral and negative sentiment. However, negative posts have a higher average engagement rate than positive posts, while neutral posts fall between them. This indicates that higher engagement is not necessarily associated with positive sentiment and that negative content can also generate substantial audience interaction.

Overall Summary:
The analysis indicates that content characteristics, user behaviour, device type, account verification, and sentiment are all relevant factors when examining social media engagement. Video and text content show relatively higher engagement rates, while Food and Tech are among the stronger-performing content categories. Age has a very weak relationship with engagement, whereas device type and verification status show some differences in engagement-related behaviour. The sentiment analysis also shows that negative content can receive higher engagement than positive content. These findings can help identify patterns in content performance and audience behaviour.
